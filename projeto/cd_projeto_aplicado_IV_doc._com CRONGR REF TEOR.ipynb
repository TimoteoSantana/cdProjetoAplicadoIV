# **Referencial Teórico**

A fundamentação teórica deste projeto estrutura-se na intersecção entre a análise de séries temporais, a econometria aplicada ao mercado de trabalho e os Objetivos de Desenvolvimento Sustentável da Organização das Nações Unidas, em especial o ODS-8 (ONU, 2015).

## **1. O Mercado de Trabalho Formal e o Novo CAGED**
O monitoramento do emprego formal no Brasil é historicamente fundamentado nos registros administrativos consolidados pelo Ministério do Trabalho e Emprego por meio do Novo Cadastro Geral de Empregados e Desempregados (Novo CAGED), que integra informações do eSocial, do CAGED tradicional e do Empregador Web (Brasil, 2026). O saldo mensal de vagas reflete um registro administrativo sujeito a revisões metodológicas e variações sazonais (Brasil, 2026). A construção de um modelo preditivo para este indicador não substitui a divulgação oficial, mas atua como um sistema de alerta precoce para antecipar tendências de expansão ou retração setorial, subsidiando políticas ativas de emprego e o planejamento de instituições de qualificação profissional.

## **2. Conceitos e Decomposição de Séries Temporais**
Uma série temporal consiste em uma sequência de observações numéricas ordenadas cronologicamente e coletadas em intervalos regulares (Hyndman; Athanasopoulos, 2018). Sob a ótica clássica, a série pode ser decomposta em quatro componentes fundamentais (Silva; Peres; Boscarioli, 2016):
* **Tendência ($T_t$):** Direção de longo prazo da série (crescente, decrescente ou estável).
* **Sazonalidade ($S_t$):** Padrões de flutuação periódicos e regulares vinculados a ciclos fixos de calendário.
* **Ciclos ($C_t$):** Variações oscilatórias de médio/longo prazo associadas a ciclos macroeconômicos, sem periodicidade fixa.
* **Resíduo ou Ruído ($\epsilon_t$):** Variações aleatórias e irregulares não explicadas pelas demais componentes.

Adicionalmente, propriedades como estacionariedade (constância de média e variância ao longo do tempo) e autocorrelação orientam a escolha das técnicas de modelagem.

## **3. Trabalhos Correlacionados e Uso de Variáveis Exógenas**
Na literatura recente, abordagens de previsão econômica têm explorado o enriquecimento de modelos temporais com dados de alta frequência e indicadores de intenção. Em estudo correlato, Gama e Arruda (2024) demonstraram que a combinação de séries administrativas do CAGED com indicadores digitais (como tendências de busca em plataformas web) melhorou a acurácia de previsão fora da amostra em relação a modelos puramente univariados. Essa evidência fundamenta a viabilidade de incorporar variáveis exógenas no pipeline preditivo, mantendo o rigor na validação por janelas temporais.

## **4. Alternativas de Modelagem: Vantagens e Limitações**
Para a previsão do saldo mensal de vagas, três principais alternativas metodológicas foram avaliadas:
* **SARIMA (Seasonal Autoregressive Integrated Moving Average):** Captura autocorrelações lineares e sazonalidades de forma altamente interpretável (Hyndman; Athanasopoulos, 2018). *Vantagens:* Rigor estatístico consolidado para séries estáveis. *Limitações:* Alta sensibilidade a choques estruturais severos e rigidez diante de mudanças abruptas de regime.
* **Prophet:** Desenvolvido pelo Meta, baseia-se em modelos aditivos generalizados com tendências por partes e múltiplos componentes sazonais (Taylor; Letham, 2018). *Vantagens:* Robusto a dados ausentes, feriados e quebras de tendência, com parametrização flexível. *Limitações:* Requer cautela nos parâmetros de suavização para evitar extrapolações instáveis.
* **Modelos de Aprendizado de Máquina (XGBoost, Redes LSTM):** *Vantagens:* Elevada capacidade de mapear relações não lineares complexas. *Limitações:* Exigem grande volume de dados históricos, maior esforço em engenharia de atributos e apresentam menor interpretabilidade gerencial ("caixa preta").

Diante desse panorama, o projeto adotará a comparação estruturada entre uma linha de base sazonal ingênua (*baseline*), o SARIMA e o Prophet, avaliados por métricas robustas a zeros e valores negativos (MAE, RMSE, sMAPE e MASE), selecionando o modelo que melhor concilie acurácia estatística e utilidade prática para os gestores educacionais e públicos.

## Impacto Populacional
 
Decidiu-se que as séries de admissões e desligamentos do Novo CAGED serão analizadas em valores absolutos, ainda que, à primeira vista, possa parecer lógico que se considere o impacto do crescimento populacional em cada estado como uma fonte de distorção em cada série. Porém, haja vista que o objetivo é caracterizar e decompor o colume mensal de comimentações de emprego formal em cada unidade federal, optou-se por não dividir esses fluxos oer capita pelos motivos que se elenca:
- Objetivo da análise: ainda que haja possibilidade de tentência causada (e não afirmamos que há) pelo crescimento populacional, essa tendência é parte do sinal que se almeja analisar, não algo que precise ser removido por um divisor implório.
- Descontinuidade artificial: As projeçoes populacionais disponíveis, cuja fonte é as projeções de 2024 do IBGE, têm referência anual em 1º de julho de cada ano. Atribuir um valor fixo a todos os meses de cada ano faria o denominador mudar drasticamente nos meses em que se encerra uma competência e se inicia a próxima, causando variação espúria nesses meses.
- Interpolação não observável: interpolar os dados populacionais com base em uma progressão conceitual (como a geométrica) criaria uma trajetória menal estimada da população, o que adiciona incerteza injustificada.

## 🎯 Objetivo do Estudo
Investigar a correlação entre o volume de buscas no ambiente digital (via Google Trends) e o saldo de empregos formais criados mensalmente no Brasil, registrado pelo CAGED (Cadastro Geral de Empregados e Desempregados). O intuito principal é criar um modelo preditivo capaz de antecipar a divulgação oficial do indicador macroeconômico (prática conhecida como nowcasting ou previsão do presente). [2]
## 📊 Metodologia Utilizada

*
* Base de Dados: Analisou dados retrospectivos compreendidos entre janeiro de 2010 e junho de 2019. [2]
* Indicadores Digitais: Seleção de termos de busca no Google que refletem a intenção, o comportamento e a ansiedade do mercado de trabalho (como buscas por vagas, seguro-desemprego, demissão, etc.). [2, 3]
* Modelagem: Comparação de modelos de séries temporais tradicionais (univariados, focados apenas no histórico do próprio CAGED) contra modelos multivariados enriquecidos com as variáveis de alta frequência do Google Trends. [2]
*

## 🚀 Principais Resultados

*
* Correlação Relevante: Os autores comprovaram que existe uma relação estatística sólida entre o comportamento de busca dos usuários na internet e os dados oficiais de emprego. [2]
* Ganho de Performance Preditiva: A inclusão dos indicadores de intenção digital (alta frequência) reduziu os erros de previsão e melhorou significativamente o desempenho preditivo fora da amostra quando comparado aos modelos básicos tradicionais. [2]
*

Essa abordagem demonstra como dados alternativos de internet funcionam como termômetros em tempo real para a atividade econômica real. [2, 3]
